Find Unique Element in an array

```python
def findingUnique(a):

    if(len(a) == 0):
        return 0
    left = 0
    right = 1
    unique = 0

    while(right < len(a)):
        if(a[right] == a[left]):
            right+=1
        else:
            left+=1
            a[left] = a[right]
            right+=1
            unique +=1
          
    return unique
a = [1,1,2,2,3,4,5]
print('Unique elements in an array: ',findingUnique(a))   
```

Merged two sorted array

```python
def mergeTwoSortedArray(a,b):

    c = []
    i = 0
    j = 0

    while i < len(a) and j < len(b):

        if a[i] <= b[j]:
            c.append(a[i])
            i+=1
        else:
            c.append(b[j])
            j+=1
    while i < len(a):
        c.append(a[i])
        i+=1
    while j < len(b):
        c.append(b[j])
        j+=1
    return c 
a = [4,5,6]
b = [1,2,3]

print('Merged Sorted Array: ', mergeTwoSortedArray(a,b))
```

Reverse an array

```python
def reverseArray(a):
    left = 0
    right = len(a) -1 
    temp = 0
    while left < right:

        temp = a[left]
        a[left] = a[right]
        a[right] = temp

        left +=1
        right -=1
    return a

a = [1,2,3,4,5]

print("Reverse of an array: ",reverseArray(a))

```

```python
def checkPalandrome(s):
    left = 0
    right = len(s)-1

    while left < right:
        if s[left] != s[right]:
            return False
        left +=1
        right-=1
    return True
s = "KaraK"
print("Is Palindrome", checkPalandrome(s))
                
```

```python
def moveZero(a):
    left = 0
    right = 0
    temp = 0

    while left < len(a):

        if a[left] == 0:
            break
        left +=1
    right = left +1
    while right < len(a):

        if a[right] != 0:
            a[left], a[right] = a[right], a[left]

            left+=1
            right +=1
        else:
            right+=1
            
        
            

    return a
a = [0,0,1,0,4,0,0,1]
print("Move Zeroes: ", moveZero(a))
```

```python
def threeSum(a):
    results = []
    for i in range(0,len(a)-1):
        left = i+1
        right = len(a)-1
        if i > 0 and  a[i] == a[i-1] :
            continue
        while left < right:
            total = a[i] + a[left] + a[right]
            
            if total == 0:
                results.append([a[i],a[left],a[right]])
                left+=1
                while left < right and a[left] == a[left-1]:
                    left +=1
                while left < right and a[right] == a[right+1]:
                    right +=1

            elif total > 0:
                right -=1
            else:
                left +=1
    return results
nums = [-1, -1, 0, 1, 1]
print('Total combination: ',threeSum(nums))
```

Two walls problem

```python
def twoWalls(a):
    left = 0
    right = len(a)-1
    max_area = 0
    while left < right:
        area = (right-left) * min(a[left],a[right])
        max_area = max(max_area,area)
        if a[left] < a[right]:
            left +=1
        else:
           right -=1

    return max_area

a = [21,3,12,2,213]
print("Maximum: ",twoWalls(a))
    
```

```python
import math
def findPairs():
    left = 0 
    c = 5
    right = int(math.sqrt(c))
    combination = []
    while left <= right:
        results = (left*left) + (right*right)
        if  results == c:
            combination.append([left,right])
            break
        elif results < c:
            left +=1
        else:
            right-=1

    return combination
print('Combinations',findPairs())
```
