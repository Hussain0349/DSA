```python
def findingUnique(a):

    if(len(a) == 0):
        return 0
    left = 0
    right = 1
    unique = 0

    while(right < len(a)):
        if(a[right] == a[left]):
            right+=1
        else:
            left+=1
            a[left] = a[right]
            right+=1
            unique +=1
          
    return unique
a = [1,1,2,2,3,4,5]
print('Unique elements in an array: ',findingUnique(a))   
```
