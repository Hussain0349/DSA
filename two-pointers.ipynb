```python
def findingUnique(a):

    if(len(a) == 0):
        return 0
    left = 0
    right = 1
    unique = 0

    while(right < len(a)):
        if(a[right] == a[left]):
            right+=1
        else:
            left+=1
            a[left] = a[right]
            right+=1
            unique +=1
          
    return unique
a = [1,1,2,2,3,4,5]
print('Unique elements in an array: ',findingUnique(a))   
```

Merged two sorted array

```python
def mergeTwoSortedArray(a,b):

    c = []
    i = 0
    j = 0

    while i < len(a) and j < len(b):

        if a[i] <= b[j]:
            c.append(a[i])
            i+=1
        else:
            c.append(b[j])
            j+=1
    while i < len(a):
        c.append(a[i])
        i+=1
    while j < len(b):
        c.append(b[j])
        j+=1
    return c 
a = [4,5,6]
b = [1,2,3]

print('Merged Sorted Array: ', mergeTwoSortedArray(a,b))
```
