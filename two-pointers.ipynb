```python
def findingUnique(a):

    left = 0
    right = 1
    unique = 0
    temp = 0

    while(right < len(a)):
        if(a[right] == a[left]):
            right+=1
        else:
            
            unique +=1
            temp = a[right]
            a[right] = a[left]
            a[left] = temp
            left+=1
    return unique

a = [1,1,2,2,3,4,5]
print('Unique elements in an array: ',findingUnique(a))




    
```
